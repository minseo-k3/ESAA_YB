```python
import pandas as pd
import lightgbm as lgbm
```

2. 데이터를 불러옵니다.

```python
train=pd.read_csv(r'C:\Users\user\ESAA_YB\미니프로젝트2\data\train.csv')
test = pd.read_csv(r'C:\Users\user\ESAA_YB\미니프로젝트2\data\test.csv')
smpl_sub = pd.read_csv(r'C:\Users\user\ESAA_YB\미니프로젝트2\data\sample_submission.csv')
```

```python
train.head()
```

```python
test.head()
```

3. 변수를 변환합니다.

```python
# type에는 white와 red 두 종류가 있습니다.
# 각각 0,1로 변환합니다.

train['type'] = train['type'].map({'white': 0, 'red': 1})
test['type'] = test['type'].map({'white': 0, 'red': 1})
```

```python
train.head()
```

```python
test.head()
```

```python
# 모델에 입력하기 전 데이터를 정형합니다.
# train_x : train의 'index', 'quality' 피처를 제거하고 저장합니다.
# train_y : train의 'quality'를 저장합니다.
# test_x : test의 'index'를 제거하고 저장합니다.

train_x = train.drop(['index', 'quality'], axis=1)
train_y = train['quality']
test_x = test.drop(['index'], axis=1)
```

```python
train_x.shape, train_y.shape, test_x.shape
```

```python
train_x.head()
```

```python
train_y.head()
```

```python
test_x.head()
```

4. LightGBM 모델을 생성 및 훈련합니다.

```python
model = lgbm.LGBMClassifier()
model.fit(train_x, train_y)
```

5. 생성한 모델로 예측 데이터를 생성합니다.

```python
preds = model.predict(test_x)
```

6. 제출파일을 생성합니다.

```python
smpl_sub = pd.read_csv('data/sample_submission.csv')
```

```python
#서브미션 파일의 quality 변수에 예측값을 대입합니다.
smpl_sub['quality'] = preds
```

```python
# 라이브러리 및 데이터 불러오기
import pandas as pd
from xgboost import XGBClassifier

train = pd.read_csv('C:/Users/user/ESAA_YB/미니프로젝트2/data/train.csv')
test = pd.read_csv('C:/Users/user/ESAA_YB/미니프로젝트2/data/test.csv')

train.drop('index', axis=1, inplace=True)
test.drop('index', axis=1, inplace=True)
```

# 3. 보팅

```python
# 데이터 다운로드 링크로 데이터를 코랩에 불러옵니다.

!wget 'https://bit.ly/3i4n1QB'

import zipfile
with zipfile.ZipFile('3i4n1QB', 'r') as existing_zip:
    existing_zip.extractall('data')
```

```python
# 라이브러리 및 데이터 불러오기
import pandas as pd
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import VotingClassifier

train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')
train.drop('index', axis=1, inplace=True)
test.drop('index', axis=1, inplace=True)
```

```python
# 데이터 확인
train.head()
```

```python
# 원핫 인코딩 (pd.get_dummies())

train_ohe = pd.get_dummies(train)
test_ohe = pd.get_dummies(test)
```

```python
# 모델 정의 (LightGBM, XGBoost, Random Forest)

# VotingClassifier 정의

from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.ensemble import VotingClassifier

# 분류기 정의 (호환성이 좋은 모델들로 구성)
lgbm_clf = LGBMClassifier(random_state=42)
rf_clf_1 = RandomForestClassifier(random_state=42)
rf_clf_2 = ExtraTreesClassifier(random_state=42)

# 보팅 분류기 설정
voting_clf = VotingClassifier(
    estimators=[('lgbm', lgbm_clf), ('rf1', rf_clf_1), ('rf2', rf_clf_2)],
    voting='hard'
)
```

```python
# 모델 학습

# X 는 train에서 quality 를 제외한 모든 변수
# y 는 train의 qulity 변수



# fit 메소드를 이용해 모델 학습


X = train_ohe.drop('quality', axis=1)
y = train_ohe['quality'] - 3

voting_clf.fit(X, y)
None  # 출력 오류 방지
```

```python
# predict 메소드와 원핫 인코딩한 테스트 데이터를 이용해 품질 예측

# 1. 테스트 데이터 예측
preds = voting_clf.predict(test_ohe)

```
