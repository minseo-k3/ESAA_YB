#### 17. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆)

#### 19. Normalize a 5x5 random matrix (★☆☆)
Hint : $Z = \frac{X - \overline{X}}{S}$

````python
c2=np.random.random((5,5))
c2=(c2-c2.mean()) / c2.std()
print(c2)
````
